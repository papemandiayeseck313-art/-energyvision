```python
import time
import os


# 1️⃣ Méthode naïve
def naive(fichier):
    f = open(fichier, "r")
    lignes = f.readlines()   # tout charger
    f.close()

    d = {}

    for ligne in lignes:
        ip = ligne.split(",")[0]

        if ip in d:
            d[ip] += 1
        else:
            d[ip] = 1

    return d


# 2️⃣ Méthode optimisée
def optimise(fichier):
    d = {}

    f = open(fichier, "r")

    for ligne in f:   # lire ligne par ligne
        ip = ligne.split(",")[0]
        d[ip] = d.get(ip, 0) + 1

    f.close()
    return d


# 3️⃣ Tester les fichiers
fichiers = ["logs_10k.csv", "logs_100k.csv", "logs_1M.csv"]

for fichier in fichiers:

    if os.path.exists(fichier):

        t1 = time.time()
        naive(fichier)
        t2 = time.time()

        t3 = time.time()
        optimise(fichier)
        t4 = time.time()

        print(fichier, " -> Naif:", t2 - t1, " | Optimise:", t4 - t3)

    else:
        print("Fichier absent :", fichier)
```

```python
# TP Chapitre 3 — Générer un fichier rapport (tp-chapitre3.py)
# Notions : open, with, write, encoding="utf-8", try/except (OSError).

DONNEES = [
    ("Alice", 14),
    ("Bob", 12),
    ("Claire", 16),
]

def moyenne(notes):
    """Calcule la moyenne des notes."""
    if not notes:
        return 0
    return sum(notes) / len(notes)

def generer_contenu(donnees):
    """Construit le contenu complet du rapport (en-tête, lignes, moyenne, pied)."""
    contenu = "=== RAPPORT DES NOTES ===\n\n"
    for nom, note in donnees:
        contenu += f"{nom} : {note}\n"
    notes = [note for _, note in donnees]
    contenu += f"\nMoyenne générale : {moyenne(notes):.2f}\n"
    contenu += "\n=== FIN DU RAPPORT ===\n"
    return contenu

def enregistrer_rapport(contenu, nom_fichier="rapport.txt"):
    """Écrit le rapport dans le fichier avec gestion d'erreur."""
    try:
        with open(nom_fichier, "w", encoding="utf-8") as f:
            f.write(contenu)
        print(f"Rapport généré dans {nom_fichier}")
    except OSError as e:
        print(f"Erreur lors de l'écriture de {nom_fichier} : {e}")

def main():
    contenu = generer_contenu(DONNEES)
    enregistrer_rapport(contenu)

if __name__ == "__main__":
    main()
```

```python
FICHIERS_BULLETINS = ["bulletin-1.txt", "bulletin-2.txt"]

def lire_bulletin(nom_fichier):
    resultats = []
    try:
        with open(nom_fichier, "r", encoding="utf-8") as f:
            for ligne_num, ligne in enumerate(f, start=1):
                ligne = ligne.strip()
                if not ligne:
                    continue
                try:
                    nom, note_str = ligne.split(";")
                    note = float(note_str)
                    resultats.append((nom.strip(), note))
                except ValueError:
                    print(f"Avertissement : ligne invalide dans {nom_fichier} (ligne {ligne_num}) -> {ligne}")
    except FileNotFoundError:
        print(f"Fichier {nom_fichier} introuvable")
    return resultats

def fusionner_bulletins(liste_paires):
    dict_notes = {}
    for liste in liste_paires:
        for nom, note in liste:
            if nom in dict_notes:
                dict_notes[nom].append(note)
            else:
                dict_notes[nom] = [note]
    return dict_notes

def ecrire_synthese(moyennes_par_etudiant, nom_fichier="synthese.txt"):
    try:
        with open(nom_fichier, "w", encoding="utf-8") as f:
            for nom, moyenne in moyennes_par_etudiant.items():
                f.write(f"{nom} : {moyenne:.2f}\n")
        print(f"Synthèse écrite dans {nom_fichier}")
    except OSError as e:
        print(f"Erreur lors de l'écriture de {nom_fichier} : {e}")

# Exécution TP2
toutes_notes = []
for fichier in FICHIERS_BULLETINS:
    notes = lire_bulletin(fichier)
    toutes_notes.append(notes)

dict_notes = fusionner_bulletins(toutes_notes)

moyennes = {}
for nom, notes in dict_notes.items():
    if notes:
        moyennes[nom] = sum(notes) / len(notes)

ecrire_synthese(moyennes)
```

```python

class Personne:
    def __init__(self, prenom, nom):
        self.prenom = prenom
        self.nom = nom

    def description(self):
        return f"{self.prenom} {self.nom}"


# Classe Etudiant hérite de Personne
class Etudiant(Personne):
    def __init__(self, prenom, nom, numero_etudiant):
        super().__init__(prenom, nom)
        self.numero_etudiant = numero_etudiant

    def description(self):
        return f"Etudiant : {self.prenom} {self.nom} - N°{self.numero_etudiant}"


# Classe Enseignant hérite de Personne
class Enseignant(Personne):
    def __init__(self, prenom, nom, matiere):
        super().__init__(prenom, nom)
        self.matiere = matiere

    def description(self):
        return f"Enseignant : {self.prenom} {self.nom} - Matière : {self.matiere}"


# Classe PersonnelAdministratif hérite de Personne
class PersonnelAdministratif(Personne):
    def __init__(self, prenom, nom, poste):
        super().__init__(prenom, nom)
        self.poste = poste

    def description(self):
        return f"Personnel : {self.prenom} {self.nom} - Poste : {self.poste}"


# Liste des membres
membres = [
    Etudiant("Samba", "Traore", "ETU001"),
    Etudiant("Awa", "Diop", "ETU002"),
    Enseignant("Moussa", "Fall", "Informatique"),
    PersonnelAdministratif("Ibou", "Sarr", "Secrétariat"),
]

# Parcourir et afficher
for membre in membres:
    print(membre.description())
```

```python
"""
TP 2 : Gestion d'un championnat de football
Classes : Joueur, Equipe, Rencontre, Championnat
"""

# ─────────────────────────────────────────────
# CLASSE JOUEUR
# ─────────────────────────────────────────────
class Joueur:
    """Représente un joueur de football."""

    def __init__(self, nom: str, prenom: str, numero: int, poste: str):
        self.nom = nom
        self.prenom = prenom
        self.numero = numero          # numéro de maillot
        self.poste = poste            # ex. "Gardien", "Défenseur", "Milieu", "Attaquant"
        self.buts = 0                 # buts marqués au total dans le championnat

    def marquer_but(self):
        """Incrémente le compteur de buts du joueur."""
        self.buts += 1

    def __str__(self):
        return (f"  #{self.numero:>2}  {self.prenom} {self.nom.upper()}"
                f"  [{self.poste}]  — {self.buts} but(s)")


# ─────────────────────────────────────────────
# CLASSE EQUIPE
# ─────────────────────────────────────────────
class Equipe:
    """Représente une équipe de football avec ses joueurs et ses stats."""

    def __init__(self, nom: str, ville: str):
        self.nom = nom
        self.ville = ville
        self.joueurs: list[Joueur] = []

        # Statistiques championnat
        self.points = 0
        self.victoires = 0
        self.nuls = 0
        self.defaites = 0
        self.buts_pour = 0
        self.buts_contre = 0

    # ── Gestion des joueurs ──────────────────
    def ajouter_joueur(self, joueur: Joueur):
        """Ajoute un joueur à l'équipe (max 11 joueurs de champ)."""
        if any(j.numero == joueur.numero for j in self.joueurs):
            print(f"⚠  Le numéro {joueur.numero} est déjà pris dans {self.nom}.")
        else:
            self.joueurs.append(joueur)

    def trouver_joueur(self, numero: int) -> "Joueur | None":
        """Retourne le joueur portant ce numéro, ou None."""
        for j in self.joueurs:
            if j.numero == numero:
                return j
        return None

    # ── Mise à jour stats après match ────────
    def enregistrer_resultat(self, buts_marques: int, buts_encaisses: int):
        self.buts_pour += buts_marques
        self.buts_contre += buts_encaisses
        if buts_marques > buts_encaisses:
            self.victoires += 1
            self.points += 3
        elif buts_marques == buts_encaisses:
            self.nuls += 1
            self.points += 1
        else:
            self.defaites += 1

    @property
    def difference_buts(self):
        return self.buts_pour - self.buts_contre

    # ── Affichage ────────────────────────────
    def afficher_composition(self):
        print(f"\n{'═'*45}")
        print(f"  COMPOSITION — {self.nom} ({self.ville})")
        print(f"{'─'*45}")
        for j in sorted(self.joueurs, key=lambda x: x.numero):
            print(j)
        print(f"{'═'*45}")

    def __str__(self):
        return f"{self.nom} ({self.ville})"


# ─────────────────────────────────────────────
# CLASSE RENCONTRE
# ─────────────────────────────────────────────
class Rencontre:
    """Représente un match entre deux équipes."""

    def __init__(self, equipe_domicile: Equipe, equipe_exterieur: Equipe, journee: int = 1):
        self.equipe_dom = equipe_domicile
        self.equipe_ext = equipe_exterieur
        self.journee = journee
        self.buts_dom: list[tuple[Joueur, int]] = []   # (joueur, minute)
        self.buts_ext: list[tuple[Joueur, int]] = []
        self.termine = False

    # ── Enregistrement des buts ──────────────
    def enregistrer_but(self, equipe: Equipe, numero_joueur: int, minute: int):
        """
        Enregistre un but pour l'équipe donnée.
        equipe  : l'équipe qui marque (dom ou ext)
        numero_joueur : numéro de maillot du buteur
        minute  : minute du but
        """
        if self.termine:
            print("⚠  Ce match est déjà terminé.")
            return

        joueur = equipe.trouver_joueur(numero_joueur)
        if joueur is None:
            print(f"⚠  Joueur #{numero_joueur} introuvable dans {equipe.nom}.")
            return

        joueur.marquer_but()

        if equipe is self.equipe_dom:
            self.buts_dom.append((joueur, minute))
            print(f"  ⚽ {minute}' — {joueur.prenom} {joueur.nom} ({self.equipe_dom.nom})")
        else:
            self.buts_ext.append((joueur, minute))
            print(f"  ⚽ {minute}' — {joueur.prenom} {joueur.nom} ({self.equipe_ext.nom})")

    def terminer_match(self):
        """Clôture le match et met à jour les stats des équipes."""
        if self.termine:
            print("⚠  Match déjà terminé.")
            return
        score_dom = len(self.buts_dom)
        score_ext = len(self.buts_ext)
        self.equipe_dom.enregistrer_resultat(score_dom, score_ext)
        self.equipe_ext.enregistrer_resultat(score_ext, score_dom)
        self.termine = True
        print(f"\n  Coup de sifflet final : "
              f"{self.equipe_dom.nom} {score_dom} – {score_ext} {self.equipe_ext.nom}")

    # ── Feuille de match ─────────────────────
    def afficher_feuille(self):
        score_dom = len(self.buts_dom)
        score_ext = len(self.buts_ext)
        statut = "TERMINÉ" if self.termine else "EN COURS"

        print(f"\n{'═'*52}")
        print(f"  FEUILLE DE MATCH  —  Journée {self.journee}  [{statut}]")
        print(f"{'─'*52}")
        print(f"  {self.equipe_dom.nom:<20}  {score_dom}  –  {score_ext}  "
              f"{self.equipe_ext.nom:>20}")
        print(f"{'─'*52}")

        # Buts côte à côte
        all_minutes = sorted(
            [(m, 'dom', j) for j, m in self.buts_dom] +
            [(m, 'ext', j) for j, m in self.buts_ext]
        )
        for minute, cote, joueur in all_minutes:
            if cote == 'dom':
                print(f"  ⚽ {minute:>3}'  {joueur.prenom} {joueur.nom}")
            else:
                print(f"  {'':>35}  ⚽ {minute:>3}'  {joueur.prenom} {joueur.nom}")

        print(f"{'═'*52}")


# ─────────────────────────────────────────────
# CLASSE CHAMPIONNAT  (bonus)
# ─────────────────────────────────────────────
class Championnat:
    """Gère le classement et les rencontres d'un championnat."""

    def __init__(self, nom: str, saison: str):
        self.nom = nom
        self.saison = saison
        self.equipes: list[Equipe] = []
        self.rencontres: list[Rencontre] = []

    def inscrire_equipe(self, equipe: Equipe):
        self.equipes.append(equipe)

    def programmer_match(self, dom: Equipe, ext: Equipe, journee: int) -> Rencontre:
        r = Rencontre(dom, ext, journee)
        self.rencontres.append(r)
        return r

    def afficher_classement(self):
        classees = sorted(
            self.equipes,
            key=lambda e: (e.points, e.difference_buts, e.buts_pour),
            reverse=True
        )
        print(f"\n{'═'*65}")
        print(f"  CLASSEMENT — {self.nom}  |  Saison {self.saison}")
        print(f"{'─'*65}")
        print(f"  {'#':>2}  {'Équipe':<22}  {'Pts':>4}  {'J':>3}  "
              f"{'V':>3}  {'N':>3}  {'D':>3}  {'BP':>4}  {'BC':>4}  {'DB':>4}")
        print(f"{'─'*65}")
        for rang, eq in enumerate(classees, 1):
            matchs_joues = eq.victoires + eq.nuls + eq.defaites
            db = f"+{eq.difference_buts}" if eq.difference_buts >= 0 else str(eq.difference_buts)
            print(f"  {rang:>2}  {eq.nom:<22}  {eq.points:>4}  {matchs_joues:>3}  "
                  f"{eq.victoires:>3}  {eq.nuls:>3}  {eq.defaites:>3}  "
                  f"{eq.buts_pour:>4}  {eq.buts_contre:>4}  {db:>4}")
        print(f"{'═'*65}")

    def afficher_meilleurs_buteurs(self, top: int = 5):
        tous = [j for eq in self.equipes for j in eq.joueurs]
        buteurs = sorted([j for j in tous if j.buts > 0], key=lambda j: j.buts, reverse=True)
        print(f"\n{'═'*45}")
        print(f"  TOP {top} BUTEURS — {self.nom}")
        print(f"{'─'*45}")
        for i, j in enumerate(buteurs[:top], 1):
            print(f"  {i}. {j.prenom} {j.nom.upper():<20} {j.buts} but(s)")
        print(f"{'═'*45}")


# ═════════════════════════════════════════════
# PROGRAMME PRINCIPAL — démonstration
# ═════════════════════════════════════════════
if __name__ == "__main__":

    # ── Création du championnat ──────────────
    ligue = Championnat("Ligue Pro Sénégal", "2025-2026")

    # ── Création des équipes ─────────────────
    casa = Equipe("Casa Sports", "Ziguinchor")
    generation = Equipe("Génération Foot", "Dakar")
    teungueth = Equipe("Teungueth FC", "Rufisque")

    ligue.inscrire_equipe(casa)
    ligue.inscrire_equipe(generation)
    ligue.inscrire_equipe(teungueth)

    # ── Ajout des joueurs ────────────────────
    casa.ajouter_joueur(Joueur("Diallo", "Ibrahima", 1, "Gardien"))
    casa.ajouter_joueur(Joueur("Mendy", "Cheikh", 5, "Défenseur"))
    casa.ajouter_joueur(Joueur("Sambou", "Ousmane", 10, "Milieu"))
    casa.ajouter_joueur(Joueur("Coly", "Pape", 9, "Attaquant"))
    casa.ajouter_joueur(Joueur("Badji", "Saliou", 7, "Attaquant"))

    generation.ajouter_joueur(Joueur("Gueye", "Mamadou", 1, "Gardien"))
    generation.ajouter_joueur(Joueur("Ndiaye", "Aliou", 4, "Défenseur"))
    generation.ajouter_joueur(Joueur("Sarr", "Lamine", 8, "Milieu"))
    generation.ajouter_joueur(Joueur("Fall", "Assane", 11, "Attaquant"))
    generation.ajouter_joueur(Joueur("Diouf", "El Hadji", 9, "Attaquant"))

    teungueth.ajouter_joueur(Joueur("Faye", "Babacar", 1, "Gardien"))
    teungueth.ajouter_joueur(Joueur("Thiam", "Ibou", 3, "Défenseur"))
    teungueth.ajouter_joueur(Joueur("Sow", "Moussa", 6, "Milieu"))
    teungueth.ajouter_joueur(Joueur("Diop", "Serigne", 10, "Milieu"))
    teungueth.ajouter_joueur(Joueur("Ba", "Amara", 9, "Attaquant"))

    # ── Affichage des compositions ───────────
    casa.afficher_composition()
    generation.afficher_composition()

    # ── Match 1 : Casa Sports vs Génération Foot ──
    print("\n\n>>> JOURNÉE 1 — MATCH 1")
    match1 = ligue.programmer_match(casa, generation, journee=1)

    match1.enregistrer_but(casa, 9, 23)       # Coly, 23'
    match1.enregistrer_but(generation, 9, 41) # Diouf, 41'
    match1.enregistrer_but(casa, 7, 67)       # Badji, 67'
    match1.enregistrer_but(casa, 9, 89)       # Coly, 89'

    match1.terminer_match()
    match1.afficher_feuille()

    # ── Match 2 : Génération Foot vs Teungueth FC ──
    print("\n\n>>> JOURNÉE 1 — MATCH 2")
    match2 = ligue.programmer_match(generation, teungueth, journee=1)

    match2.enregistrer_but(teungueth, 9, 12)  # Ba, 12'
    match2.enregistrer_but(generation, 11, 55) # Fall, 55'
    match2.enregistrer_but(teungueth, 10, 78) # Diop, 78'

    match2.terminer_match()
    match2.afficher_feuille()

    # ── Match 3 : Casa Sports vs Teungueth FC ──
    print("\n\n>>> JOURNÉE 2 — MATCH 3")
    match3 = ligue.programmer_match(casa, teungueth, journee=2)

    match3.enregistrer_but(teungueth, 9, 5)   # Ba, 5'
    match3.enregistrer_but(casa, 10, 38)       # Sambou, 38'
    match3.enregistrer_but(casa, 9, 72)        # Coly, 72'

    match3.terminer_match()
    match3.afficher_feuille()

    # ── Classement final ─────────────────────
    ligue.afficher_classement()

    # ── Meilleurs buteurs ────────────────────
    ligue.afficher_meilleurs_buteurs(top=5)
```

```python
"""
TP 3 : Mini-système de bibliothèque
Classes : Livre, Lecteur, Emprunt, Bibliotheque
Fonctionnalités : catalogue, emprunts, retours, historique fichier, gestion d'erreurs
"""

import os
from datetime import date, timedelta

HISTORIQUE_FICHIER = "historique_emprunts.txt"


# ─────────────────────────────────────────────
# EXCEPTIONS PERSONNALISÉES
# ─────────────────────────────────────────────

class LivreDejaEmprunte(Exception):
    """Levée quand on tente d'emprunter un livre déjà sorti."""
    pass

class LivreNonEmprunte(Exception):
    """Levée quand on tente de rendre un livre qui n'est pas emprunté."""
    pass

class LivreIntrouvable(Exception):
    """Levée quand l'ISBN demandé n'existe pas dans le catalogue."""
    pass

class LecteurIntrouvable(Exception):
    """Levée quand l'identifiant lecteur est inconnu."""
    pass


# ─────────────────────────────────────────────
# CLASSE LIVRE
# ─────────────────────────────────────────────

class Livre:
    """Représente un livre du catalogue."""

    def __init__(self, isbn: str, titre: str, auteur: str, annee: int):
        self.isbn = isbn          # identifiant unique
        self.titre = titre
        self.auteur = auteur
        self.annee = annee
        self.disponible = True    # False quand emprunté

    def __str__(self):
        dispo = "disponible" if self.disponible else "emprunté"
        return f"[{self.isbn}] « {self.titre} » — {self.auteur} ({self.annee})  [{dispo}]"


# ─────────────────────────────────────────────
# CLASSE LECTEUR
# ─────────────────────────────────────────────

class Lecteur:
    """Représente un lecteur inscrit à la bibliothèque."""

    def __init__(self, identifiant: str, nom: str, prenom: str):
        self.identifiant = identifiant
        self.nom = nom
        self.prenom = prenom

    def __str__(self):
        return f"[{self.identifiant}] {self.prenom} {self.nom.upper()}"


# ─────────────────────────────────────────────
# CLASSE EMPRUNT
# ─────────────────────────────────────────────

class Emprunt:
    """Lie un livre à un lecteur sur une période donnée."""

    DUREE_DEFAUT = 14  # jours

    def __init__(self, livre: Livre, lecteur: Lecteur,
                 date_debut: date = None, duree_jours: int = DUREE_DEFAUT):
        self.livre = livre
        self.lecteur = lecteur
        self.date_debut = date_debut or date.today()
        self.date_retour_prevue = self.date_debut + timedelta(days=duree_jours)
        self.date_retour_effectif: date | None = None   # rempli au retour

    @property
    def rendu(self) -> bool:
        return self.date_retour_effectif is not None

    @property
    def en_retard(self) -> bool:
        """True si le livre n'est pas encore rendu et que la date prévue est dépassée."""
        return not self.rendu and date.today() > self.date_retour_prevue

    @property
    def jours_retard(self) -> int:
        if not self.rendu:
            return max(0, (date.today() - self.date_retour_prevue).days)
        return max(0, (self.date_retour_effectif - self.date_retour_prevue).days)

    def __str__(self):
        statut = f"Rendu le {self.date_retour_effectif}" if self.rendu else \
                 f"À rendre avant le {self.date_retour_prevue}"
        return (f"  Livre   : {self.livre.titre}\n"
                f"  Lecteur : {self.lecteur}\n"
                f"  Début   : {self.date_debut}   {statut}")


# ─────────────────────────────────────────────
# CLASSE BIBLIOTHÈQUE
# ─────────────────────────────────────────────

class Bibliotheque:
    """Gère le catalogue, les lecteurs et les emprunts."""

    def __init__(self, nom: str, fichier_historique: str = HISTORIQUE_FICHIER):
        self.nom = nom
        self.fichier_historique = fichier_historique
        self.livres: dict[str, Livre] = {}       # isbn → Livre
        self.lecteurs: dict[str, Lecteur] = {}   # identifiant → Lecteur
        self.emprunts_actifs: list[Emprunt] = []
        self.historique: list[Emprunt] = []      # tous les emprunts (actifs + terminés)

    # ── Gestion du catalogue ─────────────────
    def ajouter_livre(self, livre: Livre):
        self.livres[livre.isbn] = livre
        print(f"  ✔  Livre ajouté : {livre.titre}")

    def catalogue(self):
        print(f"\n{'═'*60}")
        print(f"  CATALOGUE — {self.nom}")
        print(f"{'─'*60}")
        if not self.livres:
            print("  (catalogue vide)")
        for livre in self.livres.values():
            print(f"  {livre}")
        print(f"{'═'*60}")

    def _chercher_livre(self, isbn: str) -> Livre:
        if isbn not in self.livres:
            raise LivreIntrouvable(f"ISBN « {isbn} » introuvable dans le catalogue.")
        return self.livres[isbn]

    # ── Gestion des lecteurs ─────────────────
    def inscrire_lecteur(self, lecteur: Lecteur):
        self.lecteurs[lecteur.identifiant] = lecteur
        print(f"  ✔  Lecteur inscrit : {lecteur}")

    def _chercher_lecteur(self, identifiant: str) -> Lecteur:
        if identifiant not in self.lecteurs:
            raise LecteurIntrouvable(f"Lecteur « {identifiant} » introuvable.")
        return self.lecteurs[identifiant]

    # ── Emprunts ─────────────────────────────
    def emprunter(self, isbn: str, identifiant_lecteur: str,
                  date_debut: date = None, duree_jours: int = Emprunt.DUREE_DEFAUT) -> Emprunt:
        """
        Enregistre un emprunt.
        Lève LivreIntrouvable, LecteurIntrouvable ou LivreDejaEmprunte.
        """
        livre = self._chercher_livre(isbn)
        lecteur = self._chercher_lecteur(identifiant_lecteur)

        if not livre.disponible:
            raise LivreDejaEmprunte(
                f"« {livre.titre} » est déjà emprunté. Retour prévu le "
                f"{self._emprunt_actif(isbn).date_retour_prevue}."
            )

        emprunt = Emprunt(livre, lecteur, date_debut, duree_jours)
        livre.disponible = False
        self.emprunts_actifs.append(emprunt)
        self.historique.append(emprunt)

        self._ecrire_historique(
            f"EMPRUNT  | {emprunt.date_debut} | {livre.isbn} | "
            f"{livre.titre} | {lecteur.identifiant} | {lecteur.prenom} {lecteur.nom} | "
            f"Retour prévu : {emprunt.date_retour_prevue}"
        )

        print(f"  ✔  Emprunt enregistré : « {livre.titre} » → {lecteur}")
        return emprunt

    def rendre(self, isbn: str, date_retour: date = None) -> Emprunt:
        """
        Enregistre le retour d'un livre.
        Lève LivreIntrouvable ou LivreNonEmprunte.
        Signale un retard si applicable.
        """
        livre = self._chercher_livre(isbn)
        emprunt = self._emprunt_actif(isbn)

        if emprunt is None:
            raise LivreNonEmprunte(
                f"« {livre.titre} » ne figure pas dans les emprunts actifs."
            )

        date_retour = date_retour or date.today()
        emprunt.date_retour_effectif = date_retour
        livre.disponible = True
        self.emprunts_actifs.remove(emprunt)

        retard = emprunt.jours_retard
        statut_retard = f"EN RETARD ({retard} jour(s))" if retard > 0 else "Dans les délais"

        self._ecrire_historique(
            f"RETOUR   | {date_retour} | {livre.isbn} | "
            f"{livre.titre} | {emprunt.lecteur.identifiant} | "
            f"{emprunt.lecteur.prenom} {emprunt.lecteur.nom} | {statut_retard}"
        )

        if retard > 0:
            print(f"  ⚠  Retour EN RETARD de {retard} jour(s) : « {livre.titre} »")
        else:
            print(f"  ✔  Retour enregistré : « {livre.titre} »  ({statut_retard})")

        return emprunt

    def _emprunt_actif(self, isbn: str) -> "Emprunt | None":
        """Retourne l'emprunt actif d'un livre ou None."""
        for e in self.emprunts_actifs:
            if e.livre.isbn == isbn:
                return e
        return None

    # ── Affichages ────────────────────────────
    def emprunts_en_cours(self):
        print(f"\n{'═'*60}")
        print(f"  EMPRUNTS EN COURS — {self.nom}")
        print(f"{'─'*60}")
        if not self.emprunts_actifs:
            print("  Aucun emprunt en cours.")
        for e in self.emprunts_actifs:
            retard = f"  ⚠  RETARD : {e.jours_retard} jour(s)" if e.en_retard else ""
            print(e)
            if retard:
                print(retard)
            print()
        print(f"{'═'*60}")

    def emprunts_en_retard(self):
        retards = [e for e in self.emprunts_actifs if e.en_retard]
        print(f"\n{'═'*60}")
        print(f"  EMPRUNTS EN RETARD")
        print(f"{'─'*60}")
        if not retards:
            print("  Aucun retard. 🎉")
        for e in retards:
            print(f"  {e.lecteur}  →  « {e.livre.titre} »  "
                  f"({e.jours_retard} jour(s) de retard)")
        print(f"{'═'*60}")

    def afficher_historique(self):
        print(f"\n{'═'*60}")
        print(f"  HISTORIQUE COMPLET ({len(self.historique)} emprunt(s))")
        print(f"{'─'*60}")
        for e in self.historique:
            statut = "Rendu" if e.rendu else "En cours"
            print(f"  [{statut}] {e.livre.titre}  ←→  {e.lecteur.prenom} {e.lecteur.nom}")
        print(f"{'═'*60}")

    # ── Fichier historique ────────────────────
    def _ecrire_historique(self, ligne: str):
        """Ajoute une ligne horodatée dans le fichier historique (UTF-8)."""
        try:
            with open(self.fichier_historique, "a", encoding="utf-8") as f:
                f.write(ligne + "\n")
        except OSError as e:
            print(f"  ✗  Impossible d'écrire dans l'historique : {e}")

    def lire_historique_fichier(self):
        """Affiche le contenu brut du fichier historique."""
        print(f"\n{'═'*60}")
        print(f"  FICHIER HISTORIQUE : {self.fichier_historique}")
        print(f"{'─'*60}")
        try:
            with open(self.fichier_historique, "r", encoding="utf-8") as f:
                contenu = f.read().strip()
                print(contenu if contenu else "  (fichier vide)")
        except FileNotFoundError:
            print("  (fichier introuvable — aucun emprunt enregistré)")
        except OSError as e:
            print(f"  ✗  Erreur de lecture : {e}")
        print(f"{'═'*60}")


# ═════════════════════════════════════════════
# PROGRAMME PRINCIPAL — démonstration
# ═════════════════════════════════════════════
if __name__ == "__main__":

    # Nettoyage du fichier historique pour la démo
    if os.path.exists(HISTORIQUE_FICHIER):
        os.remove(HISTORIQUE_FICHIER)

    # ── Création de la bibliothèque ──────────
    bib = Bibliotheque("Bibliothèque Centrale de Dakar")

    # ── Ajout de livres ──────────────────────
    print("\n=== Ajout des livres ===")
    bib.ajouter_livre(Livre("ISBN-001", "Les Misérables", "Victor Hugo", 1862))
    bib.ajouter_livre(Livre("ISBN-002", "Le Petit Prince", "Antoine de Saint-Exupéry", 1943))
    bib.ajouter_livre(Livre("ISBN-003", "L'Étranger", "Albert Camus", 1942))
    bib.ajouter_livre(Livre("ISBN-004", "Bonjour Tristesse", "Françoise Sagan", 1954))

    # ── Inscription des lecteurs ──────────────
    print("\n=== Inscription des lecteurs ===")
    bib.inscrire_lecteur(Lecteur("L001", "Diallo", "Aminata"))
    bib.inscrire_lecteur(Lecteur("L002", "Ndiaye", "Ibrahima"))
    bib.inscrire_lecteur(Lecteur("L003", "Sow", "Fatou"))

    # ── Affichage du catalogue ────────────────
    bib.catalogue()

    # ── Emprunts normaux ──────────────────────
    print("\n=== Emprunts ===")
    bib.emprunter("ISBN-001", "L001")
    bib.emprunter("ISBN-002", "L002", duree_jours=7)
    bib.emprunter("ISBN-003", "L003")

    # ── Tentative d'emprunt d'un livre déjà sorti ──
    print("\n=== Test : livre déjà emprunté ===")
    try:
        bib.emprunter("ISBN-001", "L002")
    except LivreDejaEmprunte as e:
        print(f"  ✗  Erreur : {e}")

    # ── Tentative avec ISBN inconnu ───────────
    print("\n=== Test : ISBN inconnu ===")
    try:
        bib.emprunter("ISBN-999", "L001")
    except LivreIntrouvable as e:
        print(f"  ✗  Erreur : {e}")

    # ── Tentative avec lecteur inconnu ────────
    print("\n=== Test : lecteur inconnu ===")
    try:
        bib.emprunter("ISBN-004", "L999")
    except LecteurIntrouvable as e:
        print(f"  ✗  Erreur : {e}")

    # ── Emprunts en cours ─────────────────────
    bib.emprunts_en_cours()

    # ── Retour normal ─────────────────────────
    print("\n=== Retour d'un livre ===")
    bib.rendre("ISBN-002")

    # ── Simulation d'un retour en retard ──────
    print("\n=== Simulation retour en retard ===")
    # On crée un emprunt avec une date de début passée (-20 jours)
    bib.emprunter(
        "ISBN-004", "L001",
        date_debut=date.today() - timedelta(days=20),
        duree_jours=14
    )
    bib.rendre("ISBN-004")    # rendu aujourd'hui → 6 jours de retard

    # ── Tentative de rendre un livre non emprunté ──
    print("\n=== Test : rendre un livre non emprunté ===")
    try:
        bib.rendre("ISBN-002")
    except LivreNonEmprunte as e:
        print(f"  ✗  Erreur : {e}")

    # ── Vérification des retards restants ─────
    bib.emprunts_en_retard()

    # ── Catalogue mis à jour ──────────────────
    bib.catalogue()

    # ── Historique en mémoire ─────────────────
    bib.afficher_historique()

    # ── Contenu du fichier historique ─────────
    bib.lire_historique_fichier()
    
```

```python
"""
Chapitre 4 — TP 1 : Animaux, zoo et rapport

Compléter les classes et le bloc de test selon l'énoncé.
"""


class Animal:
    def cri(self):
        return ""

    def __str__(self):
        return "Je suis un animal"


class Chat(Animal):
    def cri(self):
        return "Miaaaw !"

    def __str__(self):
        return "Je suis un chat !"


class Chien(Animal):
    def __str__(self):
        return "Je suis un chien !"

    def aboyer(self):
        return "Haw haw !"


class Zoo:
    def __init__(self):
        self.animaux = []

    def ajouter(self, animal):
        self.animaux.append(animal)

    def sauver_rapport(self, chemin):
        with open(chemin, "w", encoding="utf-8") as f:
            for animal in self.animaux:
                f.write(str(animal) + "\n")


if __name__ == "__main__":
    animal = Animal()
    chat = Chat()
    chien = Chien()

    print("=== Affichage des animaux ===")
    print(animal)
    print(chat)
    print(chien)

    print("\n=== Cris des animaux ===")
    print("Animal :", repr(animal.cri()))
    print("Chat   :", repr(chat.cri()))
    print("Chien  :", repr(chien.cri()))

    print("\n=== Méthode spécifique du chien ===")
    print("Chien aboie :", chien.aboyer())

    print("\n=== Zoo et rapport fichier ===")
    zoo = Zoo()
    zoo.ajouter(animal)
    zoo.ajouter(chat)
    zoo.ajouter(chien)
    zoo.sauver_rapport("zoo_rapport.txt")
    print("Rapport enregistré dans zoo_rapport.txt")

    print("\n=== Polymorphisme ===")
    a1: Animal = Chat()
    a2: Animal = Chien()
    print("a1 (Chat affecté en Animal) :", a1)
    print("a1.cri() :", repr(a1.cri()))
    print("a2 (Chien affecté en Animal) :", a2)
    print("a2.cri() :", repr(a2.cri()))
```

```python
"""
Chapitre 4 — TP 2 : Formes 2D et catalogue fichier

Compléter les classes et la fonction selon l'énoncé.
"""

PI = 3.141592653589793


class Forme2D:
    def __init__(self, couleur="noir"):
        self.couleur = couleur

    def aire(self):
        return 0.0

    def perimetre(self):
        return 0.0


class Rectangle(Forme2D):
    def __init__(self, couleur, longueur, largeur):
        super().__init__(couleur)
        self.longueur = longueur
        self.largeur = largeur

    def aire(self):
        return self.longueur * self.largeur

    def perimetre(self):
        return 2 * (self.longueur + self.largeur)


class Disque(Forme2D):
    def __init__(self, couleur, rayon):
        super().__init__(couleur)
        self.rayon = rayon

    def diametre(self):
        return 2 * self.rayon

    def aire(self):
        return PI * self.rayon ** 2

    def perimetre(self):
        return 2 * PI * self.rayon


def sauver_catalogue(formes, chemin):
    """Écrit dans chemin une ligne par forme : couleur ; aire ; perimetre."""
    with open(chemin, "w", encoding="utf-8") as f:
        for forme in formes:
            f.write(f"{forme.couleur} ; {forme.aire():.4f} ; {forme.perimetre():.4f}\n")


if __name__ == "__main__":
    formes = [
        Rectangle("rouge", 2.0, 3.0),
        Rectangle("vert", 5.0, 1.5),
        Disque("bleu", 1.0),
        Disque("jaune", 2.5),
    ]

    for f in formes:
        print(f"\nForme de couleur {f.couleur}:")
        print("  Aire      :", f.aire())
        print("  Périmètre :", f.perimetre())
        if isinstance(f, Disque):
            print("  Diamètre  :", f.diametre())

    sauver_catalogue(formes, "catalogue_formes.txt")
    print("\nCatalogue enregistré dans catalogue_formes.txt")
```

```python
import os

# ============================================================
#  CLASSES
# ============================================================

class Produit:
    def __init__(self, nom, quantite):
        self.nom = nom
        self.quantite = quantite

    def __str__(self):
        return f"{self.nom} : {self.quantite}"


class StockError(Exception):
    pass

class QuantiteInvalideError(StockError):
    pass

class StockInsuffisantError(StockError):
    pass

class Stock:
    def __init__(self):
        self._produits = []

    def get(self, nom):
        for p in self._produits:
            if p.nom == nom:
                return p
        return None

    def ajouter(self, nom, quantite):
        if not nom or nom.strip() == "":
            raise QuantiteInvalideError("Le nom du produit ne peut pas être vide.")
        if quantite < 0:
            raise QuantiteInvalideError(f"La quantité ne peut pas être négative (reçu : {quantite}).")
        produit = self.get(nom)
        if produit:
            produit.quantite += quantite
        else:
            self._produits.append(Produit(nom, quantite))

    def sortir(self, nom, quantite):
        if quantite <= 0:
            raise QuantiteInvalideError(f"La quantité à sortir doit être strictement positive (reçu : {quantite}).")
        produit = self.get(nom)
        if produit is None:
            raise StockInsuffisantError(f"Produit '{nom}' introuvable dans le stock.")
        if produit.quantite < quantite:
            raise StockInsuffisantError(
                f"Stock insuffisant pour '{nom}' : disponible {produit.quantite}, demandé {quantite}."
            )
        produit.quantite -= quantite

    def produits(self):
        return list(self._produits)

    def __str__(self):
        if not self._produits:
            return "(stock vide)"
        return "\n".join(str(p) for p in self._produits)


# ============================================================
#  SERVICES
# ============================================================

CHEMIN_STOCK = "data/stock.txt"

def charger_stock_texte(chemin):
    stock = Stock()
    try:
        with open(chemin, "r", encoding="utf-8") as f:
            for ligne in f:
                ligne = ligne.strip()
                if not ligne:
                    continue
                parties = ligne.split(";")
                if len(parties) != 2:
                    continue
                nom, qte_str = parties[0].strip(), parties[1].strip()
                if not nom:
                    continue
                try:
                    quantite = int(qte_str)
                except ValueError:
                    continue
                if quantite < 0:
                    continue
                try:
                    stock.ajouter(nom, quantite)
                except Exception:
                    continue
    except FileNotFoundError:
        pass
    return stock

def sauver_stock_texte(stock, chemin):
    with open(chemin, "w", encoding="utf-8") as f:
        for produit in stock.produits():
            f.write(f"{produit.nom};{produit.quantite}\n")


# ============================================================
#  MENU PRINCIPAL
# ============================================================

def afficher_menu():
    print("\n===== Gestion de stock =====")
    print("1. Afficher le stock")
    print("2. Ajouter un produit")
    print("3. Sortir un produit")
    print("4. Sauvegarder")
    print("5. Quitter")
    print("============================")

def saisir_int(message):
    while True:
        try:
            return int(input(message))
        except ValueError:
            print("Veuillez entrer un nombre entier valide.")

def main():
    os.makedirs("data", exist_ok=True)
    stock = charger_stock_texte(CHEMIN_STOCK)
    print("Stock chargé depuis", CHEMIN_STOCK)

    while True:
        afficher_menu()
        choix = input("Votre choix : ").strip()

        if choix == "1":
            print("\n--- Stock actuel ---")
            print(stock)

        elif choix == "2":
            nom = input("Nom du produit : ").strip()
            quantite = saisir_int("Quantité à ajouter : ")
            try:
                stock.ajouter(nom, quantite)
                print(f"Produit '{nom}' ajouté/mis à jour avec succès.")
            except StockError as e:
                print(f"Erreur : {e}")

        elif choix == "3":
            nom = input("Nom du produit : ").strip()
            quantite = saisir_int("Quantité à sortir : ")
            try:
                stock.sortir(nom, quantite)
                print(f"Sortie de {quantite} unité(s) de '{nom}' enregistrée.")
            except StockError as e:
                print(f"Erreur : {e}")

        elif choix == "4":
            sauver_stock_texte(stock, CHEMIN_STOCK)
            print(f"Stock sauvegardé dans {CHEMIN_STOCK}.")

        elif choix == "5":
            sauver_stock_texte(stock, CHEMIN_STOCK)
            print("Stock sauvegardé. Au revoir !")
            break

        else:
            print("Choix invalide, veuillez entrer un nombre entre 1 et 5.")

main()s
```

```python
import numpy as np

# Définition des matrices de travail de dimension 2
B = np.array([[1, 2], [3, 4]])
C = np.array([[2, 1], [1, 3]])

# ============================================
# 1. Rang de la matrice B
# ============================================
rang_B = np.linalg.matrix_rank(B)
print(f"Rang de B : {rang_B}")

# ============================================
# 2. Valeurs propres de la matrice C
# ============================================
valeurs_propres_C, vecteurs_propres_C = np.linalg.eig(C)
print(f"\nValeurs propres de C : {valeurs_propres_C}")
print(f"Vecteurs propres de C :\n{vecteurs_propres_C}")

# ============================================
# 3. Vérification définie positive
#    (toutes les valeurs propres > 0 ?)
# ============================================
est_definie_positive = np.all(valeurs_propres_C > 0)
print(f"\nC est définie positive : {est_definie_positive}")

# ============================================
# 4. Vérification via les mineurs principaux
#    (critère de Sylvester)
# ============================================
mineur_1 = C[0, 0]                        # Mineur d'ordre 1
mineur_2 = np.linalg.det(C)              # Mineur d'ordre 2 = det(C)

print(f"\nCritère de Sylvester :")
print(f"  Mineur principal 1 (C[0,0]) : {mineur_1}")
print(f"  Mineur principal 2 (det(C)) : {mineur_2:.4f}")
print(f"  Sylvester validé : {mineur_1 > 0 and mineur_2 > 0}")

# ============================================
# 5. Matrice D = -C
# ============================================
D = -C
print(f"\nMatrice D = -C :\n{D}")

valeurs_propres_D, _ = np.linalg.eig(D)
print(f"Valeurs propres de D : {valeurs_propres_D}")

est_definie_negative = np.all(valeurs_propres_D < 0)
est_definie_positive_D = np.all(valeurs_propres_D > 0)

if est_definie_negative:
    print("Nature de D : Définie NÉGATIVE")
elif est_definie_positive_D:
    print("Nature de D : Définie POSITIVE")
else:
    print("Nature de D : Indéfinie")
```
